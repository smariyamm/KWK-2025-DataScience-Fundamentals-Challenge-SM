```python
import csv
import matplotlib.pyplot as plt

rows = []

with open("Datasets/Attack_Dataset.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        # print(row)
        rows.append(row)

# print(rows)

rows = rows[1:] # removes headers, unnessary row
dic = {} # keep track of unique attack typea and their counts
# print(rows[0])

for row in rows:
    # print(row[3].strip())
    if row[3].strip() not in dic: # row[3] hols attack types
        dic[row[3].strip()] = 0 # cleans data, removes any extra whitespace
    dic[row[3].strip()] += 1 # keep track of # of occurence


# for key, value in dic.items():
#     print(key, ":", value)
dic['other'] = 0 # cleans irrelavent data
keys = []
for key, value in dic.items():
    if value <= 30 : # if an attack occurs 30 times or less, its too rare to be most common
        dic['other'] += 1
        keys.append(key)

for key in keys: # remove rare attack types
    del dic[key]
del dic['other']


# plot using mathplotlib
print(dic) 
plt.figure(figsize=(20, 6)) # width=20, height=6
plt.bar(dic.keys(),dic.values())
plt.xlabel("Attack Type")
plt.ylabel("Count")
plt.title("Most Common Attack Types")
plt.xticks(rotation=45, ha="right", fontsize=11)
plt.tight_layout()
plt.show()




```

```python
import csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV into DataFrame
rows = []
with open("Datasets/Global_Cybersecurity_Threats_2015-2024.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        rows.append(row)

header = rows[0]
rows = rows[1:]  # remove header

# Create DataFrame
df = pd.DataFrame(rows, columns=header) 

# clean data, remove whitespace
df["Attack Type"] = df["Attack Type"].str.strip()
df["Security Vulnerability Type"] = df["Security Vulnerability Type"].str.strip()

# pivot table for all attack types vs vulnerabilities
pivot = pd.pivot_table(df, 
                       index="Security Vulnerability Type", 
                       columns="Attack Type", 
                       values="Country",  # any column to count total value num
                       aggfunc="count", 
                       fill_value=0)


# plot heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(pivot, cmap="Reds", annot=True, fmt="d", cbar=True)
# cmap="Reds": draker red intensity = higher frequency
# annot=True: shows numbers inside cells.
# fmt="d": numbers formatted as integers.

plt.title("Connections Between Attack Types and Vulnerabilities", fontsize=16)
plt.xlabel("Attack Type", fontsize=12)
plt.ylabel("Vulnerability", fontsize=12)
plt.yticks(rotation=0, fontsize=10)
plt.xticks(rotation=45, ha="right", fontsize=10)

plt.tight_layout()
plt.show()
```
